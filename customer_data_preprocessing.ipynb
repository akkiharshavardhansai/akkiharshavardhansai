```python
import pandas as pd
import numpy as np

df = pd.DataFrame({
    'customer_id': [1, 2, 3, 4, 2, 5],
    'age':         [25, np.nan, 35, np.nan, np.nan, 40],
    'city':        ['Mumbai', 'Delhi', np.nan, 'Mumbai', 'Delhi', np.nan],
    'gender':      ['Male', 'Female', 'Female', 'Male', 'Female', 'Male'],
    'annual_income': [40000, 80000, 60000, 120000, 80000, 95000]
})

df['age'] = df['age'].fillna(df['age'].median())
df['city'] = df['city'].fillna(df['city'].mode()[0])
df = df.drop_duplicates()

print("Cleaned Data:")
print(df)
```

```python
from sklearn.preprocessing import LabelEncoder

df = pd.get_dummies(df, columns=['city'], dtype=int)

le = LabelEncoder()
df['gender'] = le.fit_transform(df['gender'])

print("Encoded Data:")
print(df)
```

```python
from sklearn.preprocessing import MinMaxScaler, StandardScaler

df_minmax = df.copy()
df_standard = df.copy()

mm = MinMaxScaler()
df_minmax[['age', 'annual_income']] = mm.fit_transform(df_minmax[['age', 'annual_income']])

print("Min-Max Scaled Data:")
print(df_minmax)

ss = StandardScaler()
df_standard[['age', 'annual_income']] = ss.fit_transform(df_standard[['age', 'annual_income']])

print("\nStandard Scaled Data:")
print(df_standard)
```

Min-Max scaling is used when values need to be between 0 and 1, especially for models like neural networks. Standardization is preferred when data has outliers or follows a normal distribution, as it centers data around mean 0.
